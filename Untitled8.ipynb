```python
#  STEP 1: Install Required Libraries
!pip install -q pandas numpy qdrant-client sentence-transformers anthropic fastapi uvicorn python-multipart

#  STEP 2: Imports & Setup
import os
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
from qdrant_client import QdrantClient
from qdrant_client.models import VectorParams, Distance, PointStruct
import anthropic
from fastapi import FastAPI
from pydantic import BaseModel
import uvicorn
from threading import Thread

#  STEP 3: Set API Keys and Configuration
os.environ["ANTHROPIC_API_KEY"] = "API_KEY"  # Replace with your Claude API key
DATA_PATH = "/content/Investment_Banking_Underwriting_Market_Risk_Assessment_Corrected_Dataset.csv"

#  STEP 4: Load Embedder Model
def load_embedder():
    return SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2", device="cpu")

embedder = load_embedder()
vector_size = embedder.get_sentence_embedding_dimension()a

#  STEP 5: Load CSV and Format RAG Documents
df = pd.read_csv(DATA_PATH)
rag_docs = []
for _, row in df.iterrows():
    text = f"""
    Issuance ID {int(row['Issuance ID'])}: The underwriting process for {row['Type of Issuance']} involves an expected amount of {row['Issuance Size']} {row['Currency']} with a market risk score of {row['Market Risk Score']}, counterparty risk score of {row['Counterparty Risk Score']}, and liquidity risk score of {row['Liquidity Risk Score']}. Compliance checks for regulations are flagged as {row['Regulatory Compliance Flags']}.
    """
    rag_docs.append(text.strip())

#  STEP 6: Initialize Qdrant Vector Store
def init_qdrant(docs):
    client = QdrantClient(":memory:")
    if not client.collection_exists("investment_risk_assessment"):
        client.create_collection(
            collection_name="investment_risk_assessment",
            vectors_config=VectorParams(size=vector_size, distance=Distance.COSINE),
        )
    points = [PointStruct(id=i, vector=embedder.encode(doc).tolist(), payload={"text": doc}) for i, doc in enumerate(docs)]
    client.upsert(collection_name="investment_risk_assessment", points=points)
    return client

qdrant = init_qdrant(rag_docs)

#  STEP 7: Search Function for Document Retrieval
def search_docs(query, k=3):
    query_vec = embedder.encode(query).tolist()
    results = qdrant.search(collection_name="investment_risk_assessment", query_vector=query_vec, limit=k)
    return [r.payload["text"] for r in results]

#  STEP 8: Claude Inference Function for Response Generation
def ask_claude(context_docs, query):
    client = anthropic.Anthropic(api_key=os.environ["ANTHROPIC_API_KEY"])
    context = "\n\n".join(context_docs)
    full_prompt = f"Context:\n{context}\n\nQuestion:\n{query}\n\nAnswer:"
    response = client.messages.create(
        model="claude-3-haiku-20240307",
        max_tokens=512,
        temperature=0.2,
        system="You are a financial compliance assistant for investment banking.",
        messages=[{"role": "user", "content": full_prompt}]
    )
    return response.content[0].text

#  STEP 9: FastAPI App Setup for Serving RAG Responses
app = FastAPI()

class QueryInput(BaseModel):
    text: str

@app.post("/query/")
async def query_rag_system(query: QueryInput):
    context = search_docs(query.text)
    answer = ask_claude(context, query.text)
    return {
        "query": query.text,
        "context": context,
        "answer": answer
    }

#  STEP 10: Start API Server (in background)
def run_server():
    uvicorn.run(app, host="0.0.0.0", port=8000)

Thread(target=run_server).start()
```
