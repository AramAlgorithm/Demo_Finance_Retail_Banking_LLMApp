```python
# STEP 1: Install Required Libraries
!pip install -q pandas numpy qdrant-client sentence-transformers anthropic fastapi uvicorn nest-asyncio

# STEP 2: Imports & Setup
import os
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
from qdrant_client import QdrantClient
from qdrant_client.models import VectorParams, Distance, PointStruct
import anthropic
from fastapi import FastAPI
from pydantic import BaseModel
import uvicorn
import nest_asyncio
from threading import Thread

nest_asyncio.apply()

# STEP 3: Set API Keys and Config
os.environ["ANTHROPIC_API_KEY"] = "API_KEY"  # Replace with your Claude API key
DATA_PATH = "/content/Transaction_Compliance_Dashboard.csv"
COLLECTION_NAME = "payments_compliance"

# STEP 4: Load Embedder
def load_embedder():
    return SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2", device="cpu")

embedder = load_embedder()
vector_size = embedder.get_sentence_embedding_dimension()

# STEP 5: Load CSV and Format RAG Documents
df = pd.read_csv(DATA_PATH)
rag_docs = []

for _, row in df.iterrows():
    text = f"""
    Transaction {row['Transaction ID']} occurred on {row['Date']} with amount {row['Amount']} {row['Currency']} via {row['Type']}.
    Status: {row['Status']}, Compliance Flag: {row['Compliance Flag']}, Risk Level: {row['Risk Level']}.
    Decision made: {row['Decision']} by reviewer {row['Reviewer ID']} during audit {row['Audit ID']} on {row['Audit Date']}.
    Incident ID {row['Incident ID']} on {row['Incident Date']} — Action Taken: {row['Action Taken']}.
    Regulatory Framework: {row['Regulation']} — Compliance Status: {row['Compliance Status']}.
    Notes: {row['Notes']}
    """
    rag_docs.append(text.strip())

# STEP 6: Initialize Qdrant Vector Store
qdrant = None

def init_qdrant(docs):
    global qdrant
    client = QdrantClient(":memory:")
    if not client.collection_exists(COLLECTION_NAME):
        client.create_collection(
            collection_name=COLLECTION_NAME,
            vectors_config=VectorParams(size=vector_size, distance=Distance.COSINE),
        )
    points = [
        PointStruct(id=i, vector=embedder.encode(doc).tolist(), payload={"text": doc})
        for i, doc in enumerate(docs)
    ]
    client.upsert(collection_name=COLLECTION_NAME, points=points)
    qdrant = client

init_qdrant(rag_docs)

# STEP 7: Search Function
def search_docs(query, k=3):
    global qdrant
    query_vec = embedder.encode(query).tolist()
    results = qdrant.search(collection_name=COLLECTION_NAME, query_vector=query_vec, limit=k)
    return [r.payload["text"] for r in results]

# STEP 8: Claude Inference Function
def ask_claude(context_docs, query):
    client = anthropic.Anthropic(api_key=os.environ["ANTHROPIC_API_KEY"])
    context = "\n\n".join(context_docs)
    full_prompt = f"""Context:\n{context}\n\nQuestion:\n{query}\n\nAnswer:"""
    response = client.messages.create(
        model="claude-3-haiku-20240307",
        max_tokens=512,
        temperature=0.2,
        system="You are a payments compliance and governance assistant.",
        messages=[{"role": "user", "content": full_prompt}]
    )
    return response.content[0].text

# STEP 9: FastAPI App Setup
app = FastAPI()

class QueryInput(BaseModel):
    text: str

@app.post("/query/")
async def query_rag_system(query: QueryInput):
    context = search_docs(query.text)
    answer = ask_claude(context, query.text)
    return {
        "query": query.text,
        "context": context,
        "answer": answer
    }

# STEP 10: Run FastAPI Server
def run_server():
    uvicorn.run(app, host="0.0.0.0", port=8000)

Thread(target=run_server).start()

# STEP 11: Interactive Query Mode
import requests

def run_query_interaction():
    try:
        print("RAG System is now ready to accept queries. Type 'exit' to stop.")
        while True:
            query_text = input("Enter your transaction compliance question:\n> ")
            if query_text.lower() == 'exit':
                print("Exiting query interaction.")
                break

            response = requests.post("http://localhost:8000/query/", json={"text": query_text})

            if response.status_code == 200:
                data = response.json()
                print("\n Query:", data["query"])
                print("\n Context Used:")
                for ctx in data["context"]:
                    print("-", ctx)
                print("\n Answer from Claude:")
                print(data["answer"])
            else:
                print("Server error:", response.status_code)
    except Exception as e:
        print("Error during interaction:", str(e))

# Run interactive query interface
if __name__ == "__main__":
    run_query_interaction()
```
