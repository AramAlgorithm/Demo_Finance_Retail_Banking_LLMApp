```python
#  STEP 1: Install Libraries
!pip install -q pandas numpy qdrant-client sentence-transformers anthropic fastapi uvicorn nest-asyncio

#  STEP 2: Imports & Setup
import os
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
from qdrant_client import QdrantClient
from qdrant_client.models import VectorParams, Distance, PointStruct
import anthropic
from fastapi import FastAPI
from pydantic import BaseModel
import uvicorn
import nest_asyncio
from threading import Thread

nest_asyncio.apply() 

#  STEP 3: Set API Keys and Configuration
os.environ["ANTHROPIC_API_KEY"] = "API_KEY"  # Replace with your Claude API key
DATA_PATH = "/content/credit_card_data.csv"
COLLECTION_NAME = "credit_card_management"

#  STEP 4: Load Embedder Model
def load_embedder():
    return SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2", device="cpu")

embedder = load_embedder()
vector_size = embedder.get_sentence_embedding_dimension()

#  STEP 5: Load CSV and Format RAG Documents
df = pd.read_csv(DATA_PATH)
rag_docs = []
for _, row in df.iterrows():
    text = f"""
    Cardholder {int(row['CardholderID'])} with account number {row['AccountNumber']} has a credit limit of ${row['CreditLimit']},
    available credit of ${row['AvailableCredit']}, total spent this month is ${row['TotalSpentThisMonth']},
    minimum payment due is ${row['MinimumPayment']}, payment status is {row['PaymentStatus']},
    interest rate at {row['InterestRate']*100}%, fees charged ${row['FeesCharged']},
    recent large transactions: {row['RecentLargeTransactions']}, risk score {row['RiskScore']},
    compliance issues: {row['ComplianceIssues']}. Last payment date was {row['LastPaymentDate']},
    next payment due date is {row['NextPaymentDueDate']}.
    """
    rag_docs.append(text.strip())

#  STEP 6: Initialize Qdrant Vector Store
def init_qdrant(docs):
    client = QdrantClient(":memory:")
    if not client.collection_exists(COLLECTION_NAME):
        client.create_collection(
            collection_name=COLLECTION_NAME,
            vectors_config=VectorParams(size=vector_size, distance=Distance.COSINE),
        )
    points = [
        PointStruct(id=i, vector=embedder.encode(doc).tolist(), payload={"text": doc})
        for i, doc in enumerate(docs)
    ]
    client.upsert(collection_name=COLLECTION_NAME, points=points)
    return client

qdrant = init_qdrant(rag_docs)

#  STEP 7: Search Function for Document Retrieval
def search_docs(query, k=3):
    query_vec = embedder.encode(query).tolist()
    results = qdrant.search(collection_name=COLLECTION_NAME, query_vector=query_vec, limit=k)
    return [r.payload["text"] for r in results]

#  STEP 8: Claude Inference Function for Response Generation
def ask_claude(context_docs, query):
    client = anthropic.Anthropic(api_key=os.environ["ANTHROPIC_API_KEY"])
    context = "\n\n".join(context_docs)
    full_prompt = f"""Context:\n{context}\n\nQuestion:\n{query}\n\nAnswer:"""
    response = client.messages.create(
        model="claude-3-haiku-20240307",
        max_tokens=512,
        temperature=0.2,
        system="You are a financial compliance assistant for a retail bank.",
        messages=[{"role": "user", "content": full_prompt}]
    )
    return response.content[0].text

#  STEP 9: FastAPI App Setup for Serving RAG Responses
app = FastAPI()

class QueryInput(BaseModel):
    text: str

@app.post("/query/")
async def query_rag_system(query: QueryInput):
    context = search_docs(query.text)
    answer = ask_claude(context, query.text)
    return {
        "query": query.text,
        "context": context,
        "answer": answer
    }

#  STEP 10: Start API Server (in background)
def run_server():
    uvicorn.run(app, host="0.0.0.0", port=8000)

Thread(target=run_server).start()
```

```python
# STEP 12: Interactive Testing of the RAG Application
import requests

def run_query_interaction():
    try:
        print("RAG System is now ready to accept queries. Type 'exit' to stop.")
        while True:
            query_text = input("Enter your financial compliance question:\n> ")
            if query_text.lower() == 'exit':
                print("Exiting query interaction.")
                break

            # Prepare the request payload
            payload = {'text': query_text}
            # Make a request to the FastAPI server
            response = requests.post("http://localhost:8000/query/", json=payload)

            if response.status_code == 200:
                response_data = response.json()
                print("\n Query:", response_data["query"])
                print("\n Context Used:")
                for context in response_data["context"]:
                    print("-", context)
                print("\n Answer from Claude:")
                print(response_data["answer"])
            else:
                print("Failed to retrieve response from the server. Status Code:", response.status_code)
    except Exception as e:
        print("An error occurred during the query interaction:", str(e))

# Run the query interaction function
if __name__ == "__main__":
    run_query_interaction()
```
